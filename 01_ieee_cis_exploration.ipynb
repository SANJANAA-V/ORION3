```python
import pandas as pd

train = pd.read_csv("train_transaction.csv")
identity = pd.read_csv("train_identity.csv")
```

```python
data = train.merge(identity, on="TransactionID", how="left")
```

```python
data.shape
```

```python
data.memory_usage(deep=True).sum() / 1024**2
```

```python
data.dtypes.value_counts()
```

```python
# STEP 1 — Check memory BEFORE reduction

before_memory = data.memory_usage(deep=True).sum() / 1024**2

print("Memory BEFORE:", round(before_memory, 2), "MB")
print("\nData types BEFORE:")
print(data.dtypes.value_counts())


# STEP 2 — Find numeric columns

float_cols = data.select_dtypes(include=["float64"]).columns
int_cols = data.select_dtypes(include=["int64"]).columns


# STEP 3 — Downcast numeric columns

data[float_cols] = data[float_cols].astype("float32")
data[int_cols] = data[int_cols].astype("int32")


# STEP 4 — Check memory AFTER reduction

after_memory = data.memory_usage(deep=True).sum() / 1024**2

print("\nMemory AFTER:", round(after_memory, 2), "MB")
print("\nData types AFTER:")
print(data.dtypes.value_counts())


# STEP 5 — Show how much memory we saved

saved = before_memory - after_memory

print("\nMemory SAVED:", round(saved, 2), "MB")
print("Reduction:", round((saved / before_memory) * 100, 2), "%")
```

```python
# STEP 2 — TIME-BASED TRAIN / VALIDATION SPLIT

# 1. Sort transactions from earliest → latest
data = data.sort_values("TransactionDT").reset_index(drop=True)

# 2. Find the 80% split point
split_idx = int(len(data) * 0.80)

# 3. First 80% = training
train_df = data.iloc[:split_idx].copy()

# 4. Last 20% = validation
valid_df = data.iloc[split_idx:].copy()

# 5. Check sizes
print("Total rows:", len(data))
print("Training rows:", len(train_df))
print("Validation rows:", len(valid_df))

# 6. Verify that validation really comes AFTER training
print("\nLatest training TransactionDT:", train_df["TransactionDT"].max())
print("Earliest validation TransactionDT:", valid_df["TransactionDT"].min())
```

```python
# STEP 3 — DUMMY BASELINE

from sklearn.dummy import DummyClassifier
from sklearn.metrics import average_precision_score

# Separate the target
y_train = train_df["isFraud"]
y_valid = valid_df["isFraud"]

# Dummy model
dummy = DummyClassifier(strategy="prior")

# Dummy needs some X input, but it completely ignores it
dummy.fit(train_df[["TransactionDT"]], y_train)

# Get fraud probabilities for validation data
dummy_probs = dummy.predict_proba(
    valid_df[["TransactionDT"]]
)[:, 1]

# Calculate PR-AUC
dummy_pr_auc = average_precision_score(
    y_valid,
    dummy_probs
)

print("Actual validation fraud rate:",
      round(y_valid.mean(), 4))

print("Dummy PR-AUC:",
      round(dummy_pr_auc, 4))
```

```python
import numpy as np

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    average_precision_score,
    roc_auc_score,
    precision_score,
    recall_score
)

# -----------------------------
# 1. Select features
# -----------------------------

features = ["TransactionAmt", "ProductCD", "card4", "card6"]

X_train = train_df[features].copy()
X_valid = valid_df[features].copy()

y_train = train_df["isFraud"]
y_valid = valid_df["isFraud"]


# -----------------------------
# 2. Log-transform transaction amount
# -----------------------------

X_train["TransactionAmt"] = np.log1p(X_train["TransactionAmt"])
X_valid["TransactionAmt"] = np.log1p(X_valid["TransactionAmt"])


# -----------------------------
# 3. Separate numeric & categorical columns
# -----------------------------

numeric_features = ["TransactionAmt"]
categorical_features = ["ProductCD", "card4", "card6"]


# -----------------------------
# 4. Numeric preprocessing
# -----------------------------

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(
        strategy="median",
        add_indicator=True
    )),
    ("scaler", StandardScaler())
])


# -----------------------------
# 5. Categorical preprocessing
# -----------------------------

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])


# -----------------------------
# 6. Combine preprocessing
# -----------------------------

preprocessor = ColumnTransformer([
    ("num", numeric_pipeline, numeric_features),
    ("cat", categorical_pipeline, categorical_features)
])


# -----------------------------
# 7. Build model pipeline
# -----------------------------

model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", LogisticRegression(max_iter=1000))
])


# -----------------------------
# 8. Train ONLY on training data
# -----------------------------

model.fit(X_train, y_train)


# -----------------------------
# 9. Get fraud probabilities
# -----------------------------

valid_probs = model.predict_proba(X_valid)[:, 1]


# -----------------------------
# 10. Calculate metrics
# -----------------------------

pr_auc = average_precision_score(y_valid, valid_probs)
roc_auc = roc_auc_score(y_valid, valid_probs)

threshold = 0.5
valid_preds = (valid_probs >= threshold).astype(int)

precision = precision_score(y_valid, valid_preds, zero_division=0)
recall = recall_score(y_valid, valid_preds, zero_division=0)


# -----------------------------
# 11. Print results
# -----------------------------

print("===== LOGISTIC REGRESSION BASELINE =====")
print("PR-AUC:", round(pr_auc, 4))
print("ROC-AUC:", round(roc_auc, 4))
print("Precision @ 0.5:", round(precision, 4))
print("Recall @ 0.5:", round(recall, 4))
```

```python
from google.colab import drive
drive.mount("/content/drive")

import pandas as pd

data_folder = "/content/drive/MyDrive/ORION3/data"

# Only load the columns needed for our current model
cols = [
    "TransactionDT",
    "TransactionAmt",
    "ProductCD",
    "card4",
    "card6",
    "isFraud"
]

data = pd.read_csv(
    f"{data_folder}/train_transaction.csv",
    usecols=cols
)

print("Data shape:", data.shape)
```

```python
# Sort by transaction time
data = data.sort_values("TransactionDT").reset_index(drop=True)

# First 80% = older transactions → training
# Last 20% = newer transactions → validation
split_idx = int(len(data) * 0.80)

train_df = data.iloc[:split_idx]
valid_df = data.iloc[split_idx:]

print("Train:", train_df.shape)
print("Validation:", valid_df.shape)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import average_precision_score, roc_auc_score

features = ["TransactionAmt", "ProductCD", "card4", "card6"]

X_train = train_df[features]
X_valid = valid_df[features]

y_train = train_df["isFraud"]
y_valid = valid_df["isFraud"]

numeric_features = ["TransactionAmt"]
categorical_features = ["ProductCD", "card4", "card6"]

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median"))
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("num", numeric_pipeline, numeric_features),
    ("cat", categorical_pipeline, categorical_features)
])

tree_model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", DecisionTreeClassifier(
        max_depth=6,
        random_state=42
    ))
])

# Train
tree_model.fit(X_train, y_train)

# Predict fraud probabilities
tree_probs = tree_model.predict_proba(X_valid)[:, 1]

# Evaluate
tree_pr_auc = average_precision_score(y_valid, tree_probs)
tree_roc_auc = roc_auc_score(y_valid, tree_probs)

print("Decision Tree PR-AUC:", tree_pr_auc)
print("Decision Tree ROC-AUC:", tree_roc_auc)
```
